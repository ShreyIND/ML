```python
from sklearn.datasets import make_classification
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import StackingClassifier
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
```

```python
base_estimator=[('lr',LogisticRegression()),('rf',RandomForestClassifier()),('svc',SVC(probability=True))]
```

```python
lvl1=LogisticRegression()
```

```python
stack=StackingClassifier(estimators=base_estimator,final_estimator=lvl1,cv=5)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x,y=make_classification(n_samples=20000,n_features=25,n_redundant=5,n_informative=15,n_repeated=4)
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
stack.fit(X_train,y_train)
```

```python
y_pred=stack.predict(X_test)
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
from sklearn.ensemble import VotingClassifier
```

```python
vc=VotingClassifier(estimators=base_estimator,voting='soft')
```

```python
vc_=VotingClassifier(estimators=base_estimator,voting='hard')
```

```python
vc.fit(X_train,y_train)
```

```python
vc_.fit(X_train,y_train)
```

```python
y1_pred=vc.predict(X_test)
y2_pred=vc_.predict(X_test)
```

```python
accuracy_score(y_test,y1_pred)
```

```python
accuracy_score(y_test,y2_pred)
```
