```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
```

```python
ohe=OneHotEncoder(sparse_output=False,drop='first')
```

```python
from sklearn.compose import ColumnTransformer
```

```python
df=df_train.copy()
```

```python
df.head()
```

```python
df.drop(columns=['age','course','internet_access','gender','id','exam_difficulty'],inplace=True)
```

```python
df.head()
```

```python
transformer=ColumnTransformer(transformers=[('one_hot',ohe,['study_method']),('ordinal',OrdinalEncoder(categories=[['poor','average','good']]),['sleep_quality']),('ord2',OrdinalEncoder(categories=[['low','medium','high']]),['facility_rating'])],remainder='passthrough')
```

```python
X = df.drop(columns=['exam_score'])
y = df['exam_score']

X_transformed = transformer.fit_transform(X)

print(X_transformed.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_transformed, y, test_size=0.2, random_state=42)
from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
x_train=ss.fit_transform(X_train)
x_test=ss.transform(X_test)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
model = DecisionTreeRegressor(random_state=42)
model.fit(X_train, y_train)

print("Model training complete.")
```

```python
from sklearn.metrics import mean_absolute_error, r2_score

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"R-squared (R2): {r2:.2f}")
```

```python
from sklearn.metrics import mean_absolute_error, r2_score
def evaluate_model(model, X_train, X_test, y_train, y_test):
  y_pred = model.predict(X_test)
  mae = mean_absolute_error(y_test, y_pred)
  r2 = r2_score(y_test, y_pred)
  print(f"Mean Absolute Error (MAE): {mae:.2f}")
  print(f"R-squared (R2): {r2:.2f}")
```
