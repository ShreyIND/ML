```python
import numpy as np
import pandas as pd
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import layers,Sequential
from keras.datasets import mnist
```

```python
df=mnist
```

```python
df=df.load_data()
```

```python
df
```

```python
import matplotlib.pyplot as plt
```

```python
x_train, y_train = df[0]
x_test,y_test=df[1]
print(f"Type of x_train: {type(x_train)}")
print(f"Shape of x_train: {x_train.shape}")
print(f"Data type of x_train: {x_train.dtype}")
print(f"Minimum pixel value in x_train: {x_train.min()}")
print(f"Maximum pixel value in x_train: {x_train.max()}")

print(f"\nType of y_train: {type(y_train)}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Data type of y_train: {y_train.dtype}")
print(f"Unique labels in y_train: {np.unique(y_train)}")
```

```python
model=Sequential()
```

```python
from tensorflow.keras.layers import Dense,Flatten,Conv2D,MaxPooling2D
```

```python
model.add(Conv2D(32,kernel_size=(3,3),padding='same',activation='relu',input_shape=(28,28,1)))
model.add(MaxPooling2D(pool_size=(2,2),strides=2,padding='same'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2),strides=2,padding='same'))
model.add(Flatten())
```

```python
model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```
