```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN, Dense
```

```python
text = [
    "artificial intelligence is the future",
    "machine learning is a subset of ai",
    "deep learning uses neural networks",
    "rnn is good for sequence data",
    "python is great for data science"
]
```

```python
tokenizer=Tokenizer()
```

```python
tokenizer.fit_on_texts(text)
```

```python
tokenizer.word_index
```

```python
vocab=len(tokenizer.word_index)+1
```

```python
input_sequences = []
for line in text:
    token_list = tokenizer.texts_to_sequences([line])[0]
    print(token_list)
    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)
```

```python
input_sequences
```

```python
max_len=0
for i in input_sequences:
    if len(i)>max_len:
        max_len=len(i)
print(max_len)
input_sequences=np.array(pad_sequences(input_sequences,maxlen=max_len,padding='pre'))
```

```python
input_sequences
```

```python
x=input_sequences[:,:-1]
y=input_sequences[:,-1]
```

```python
y = tf.keras.utils.to_categorical(y, num_classes=vocab)#Do the one hot encoding
```

```python
y
```

```python
model=Sequential()
model.add(Embedding(vocab,10,input_length=max_len-1))
```

```python
model.add(SimpleRNN(100))
```

```python
model.add(Dense(vocab,activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.fit(x,y,epochs=100,verbose=1)
```

```python
def complete(text,future):
  for i in range(future):
    token_list=tokenizer.texts_to_sequences([text])[0]
    token_list=pad_sequences([token_list],maxlen=max_len-1,padding='pre')
    next=np.argmax(model.predict(token_list))
    for word,index in tokenizer.word_index.items():
      if index==next:
        text+=' '+word
        break
  return text
```

```python
print(complete("artificial intelligence", 3))
print(complete("machine", 4))
print(complete("python is", 2))
```
