```python
from sklearn.datasets import make_classification
```

```python
x,y=make_classification(n_samples=2000,n_classes=2,n_features=20,n_redundant=5,n_informative=15,random_state=1)
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
y
```

```python
(y==0).sum()
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt=DecisionTreeClassifier(max_depth=1)
```

```python
from sklearn.ensemble import BaggingClassifier, RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier,HistGradientBoostingClassifier
```

```python
ada_clf=AdaBoostClassifier(estimator=dt,n_estimators=200,learning_rate=1,random_state=1)
```

```python
gbc=GradientBoostingClassifier(n_estimators=200,learning_rate=1,random_state=1,subsample=0.8,max_depth=5)
```

```python
hgbc=HistGradientBoostingClassifier(max_iter=200,learning_rate=0.1,random_state=1,max_depth=None)
```

```python
ada_clf.fit(X_train,y_train)
gbc.fit(X_train,y_train)
hgbc.fit(X_train,y_train)
```

```python
ada_clf.score(X_test,y_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
ad_pred=ada_clf.predict(X_test)
```

```python
accuracy_score(y_test,ad_pred)
```

```python
gbc.score(X_test,y_test)
```

```python
hgbc.score(X_test,y_test)
```
